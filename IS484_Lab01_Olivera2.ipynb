```python
def agente_termostato(temperatura_actual, temperatura_objetivo=22, margen=1.0):
    """
    Agente reactivo simple para el control de temperatura.

    PEAS:
    - Percepción (S): temperatura actual del ambiente.
    - Acciones (A): encender aire acondicionado, encender calefacción
      o mantener el sistema apagado.
    - Entorno (E): ambiente o habitación cuya temperatura se desea controlar.
    - Objetivo: mantener la temperatura cercana a 22 °C.
    - Medida de desempeño (P): mantener la temperatura dentro del margen
      establecido alrededor de la temperatura objetivo.
    """

    diferencia = temperatura_actual - temperatura_objetivo

    if diferencia > margen:
        return "Encender aire acondicionado"

    elif diferencia < -margen:
        return "Encender calefacción"

    else:
        return "Mantener sistema apagado"
```
